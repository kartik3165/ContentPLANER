### Working function to get all festival for that week

```python
from datetime import date, timedelta
from typing import Dict

import httpx


async def get_festival(
    api_key: str, start_date: date | None = None, end_date_days: int = 6
) -> Dict:
    if start_date is None:
        start_date = date.today()
    country = "IN"
    year = start_date.year
    end_date = start_date + timedelta(days=end_date_days)

    festival_dates = []

    async with httpx.AsyncClient() as client:
        all_holidays = await client.get(
            "https://calendarific.com/api/v2/holidays",
            params={
                "api_key": api_key,
                "country": country,
                "year": year,
            },
        )

        all_holidays.raise_for_status()

    holidays = all_holidays.json()["response"]["holidays"]

    for week_holidays in holidays:
        holiday_date = date.fromisoformat(week_holidays["date"]["iso"])

        if start_date <= holiday_date <= end_date:
            festival_dates.append(
                {
                    "name": week_holidays["name"],
                    "date": week_holidays["date"]["iso"],
                    "description": week_holidays["description"],
                }
            )

    return festival_dates


async def main():
    req = await get_festival(api_key="", end_date_days=20)

    return req


data = await main()

data
```

```python
from pathlib import Path

from firecrawl import AsyncFirecrawl
from firecrawl.v2.types import PaginationConfig


async def crawl_webpage(
    url: str,
    api_key: str,
    pagination: bool = False,
    output_file: str = "data/own_site.md",
) -> dict:

    crawler = AsyncFirecrawl(api_key=api_key)

    started = await crawler.start_crawl(
        url,
        limit=3,
    )

    status = await crawler.get_crawl_status(
        started.id,
        pagination_config=PaginationConfig(auto_paginate=pagination),
    )

    output_path = Path("data") / output_file
    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    with output_path.open("w", encoding="utf-8") as file:
        for page in status.data:
            file.write(page.markdown or "")
            file.write("\n\n---\n\n")

    return output_path
```

```python
result = await crawl_webpage(
    url="https://en.wikipedia.org/wiki/Marvel_Cinematic_Universe",
    api_key=crawl_api_key,
    output_file="data/marvel.md",
)
```

```python
async def crawl_batch_webpage(
    urls: list,
    api_key: str,
    pagination: bool = False,
    output_file: str = "data/comp/batch_site.md",
    poll_interval: int = 1,
    timeout: int = 60,
) -> dict:

    crawler = AsyncFirecrawl(api_key=api_key)

    await crawler.batch_scrape(
        urls,
        formats=["markdown"],
        poll_interval=poll_interval,
        timeout=timeout,
    )

    output_path = Path("data") / Path("comp") / output_file
    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    with output_path.open("w", encoding="utf-8") as file:
        for page in status.data:
            file.write(page.markdown or "")
            file.write("\n\n---\n\n")

    return output_path
```

```python
async def get_insta_account_info(user_name: str, api_key: str):
    from apify_client import ApifyClientAsync

    client = ApifyClientAsync(api_key)

    run_input = {
        "usernames": [user_name],
        "includeAboutSection": False,
    }

    run = await client.actor("dSCLg0C3YEZ83HzYX").call(run_input=run_input)

    dataset = client.dataset(run.default_dataset_id)

    async for item in dataset.iterate_items():
        return item

    return None
```

```python
from apify_client import ApifyClientAsync


async def get_bulk_insta_account_info(
    user_names: list[str],
    api_key: str,
    max_accounts: int = 10,
):
    if len(user_names) > max_accounts:
        raise ValueError(
            f"Maximum {max_accounts} Instagram accounts allowed per request. "
            f"Received {len(user_names)}."
        )

    client = ApifyClientAsync(api_key)

    run_input = {
        "usernames": user_names,
        "includeAboutSection": False,
    }

    run = await client.actor("dSCLg0C3YEZ83HzYX").call(run_input=run_input)

    dataset = client.dataset(run.default_dataset_id)

    results = []

    async for item in dataset.iterate_items():
        results.append(item)

    return results
```
